# OT1 — Otimização dos Horários da Engenharia de Computação

Notebook da modelagem PLIM em Pyomo para a alocação semanal de aulas, salas e horários.

## 1. Contexto e objetivo

Descrever o problema, o escopo e os critérios de sucesso da solução.

## 2. Dados de entrada

Carregar os JSONs em `data/raw/` e documentar seus campos.

## 3. Validação e pré-processamento

Validar dados, construir conjuntos e preparar parâmetros do modelo.

## 4. Formulação PLIM

Definir conjuntos, parâmetros, variáveis de decisão e função objetivo.

## 5. Restrições fortes

Implementar carga semanal, compatibilidade de ambiente, não sobreposição de sala, conflito de professor e conflito de período/turma.

## 6. Restrições suaves: S1 e S2

Modelar penalidades para janelas na agenda de cada período e de cada professor.

## 7. Solução e relatórios

Resolver com GLPK, verificar viabilidade e apresentar a grade por sala e por período.
